# YPc2 Dataset Generator

This notebook runs `scripts/generate_ypc2_layer.py` with configurable options.

```python
from pathlib import Path
import subprocess

repo_root = Path.cwd()
if repo_root.name.lower() == "notebooks":
    repo_root = repo_root.parent

# --- Output ---
out_dir = "dataset_ypc2_layer5"
n = 200
seed = 55

# --- Surface ---
surface_size = (200.0, 200.0)

# --- Lattice ---
spacing = 15.0
spacing_range = (14.0, 16.0)  # set to None to disable
grid = 10
grid_range = (1, 10)  # set to None to disable
orientation = 0.0
orientation_range = (0.0, 44.0)  # set to None to disable
lattice_angle = 15
lattice_angle_range = None#0.0, 15.0)  # set to None to disable

# --- YPc2 LDOS ---
ypc2_r0 = 5.8
ypc2_sig_r = 1.75
ypc2_ang_mix = 0.0
ypc2_center_plateau_amp = 0.2
ypc2_center_plateau_radius = 5.0
ypc2_center_plateau_sigma = 0.3

# --- Adatom on top of YPc2 ---
adatom_on_top = (2, 6)  # set to None to disable
adatom_on_top_height = 2.0
adatom_on_top_width = 20.5
adatom_on_top_offset = None  # set to None for default
adatom_on_top_jitter = None  # set to None for default

# --- Surface / defects ---
adatom_count = (1, 10)  # set to None to disable override
vacancy_count = (2, 4)  # set to None to disable override
molecule_count = (0, 4)  # random molecules in addition to lattice
step_probability = 0.5  # set to None to keep script default
step_height_layers = 1  # set to None to keep script default
roughness_sigma = 0.02  # set to None to keep script default

# --- Noise / realism ---
realistic = False
vibration_amp = (0.005, 0.02)  # set to None to disable
vibration_wavelength = (8.0, 16.0)
vibration_angle = (0.0, 180.0)
slope_x = (-0.9, 0.9)
slope_y = (-0.2, 0.2)
```

```python
# --- Bad tip (multi-tip) ---
# not sure if working!!
tip_mode = "multi"  # set to "multi" to enable
tip_sigma = (1.0, 2.5)  # Angstroms
tip_instability = (0.0, 0.3)
tip_count = (1, 3)
tip_offset_range = (0.0, 4.0)  # Angstroms
tip_z_range = (0.0, 4.0)  # Angstroms
tip_weight_range = (0.0, 0.7)
tip_z_decay = 1.0  # Angstroms
```

```python
# --- Steps & islands ---
# not sure if working!!
step_mode = "corner"  # or "corner"
step_corner_angle_range = (60.0, 120.0)  # deg
step_corner_round_radius = 0.0  # Angstroms

island_count = (1, 3)
island_min_distance = 1.0  # Angstroms
island_orientation_range = (0.0, 60.0)  # deg, set to None to disable
island_lattice_angle_range = (0.0, 30.0)  # deg, set to None to disable
edge_remove_count = (0, 3)
interior_remove_count = (0, 2)
edge_remove_edge_prob = 0.9
```

```python
cmd = ["python", str(repo_root / "scripts" / "generate_ypc2_layer.py"), "--out", out_dir, "--n", str(n), "--seed", str(seed)]

cmd += ["--spacing", str(spacing)]

if surface_size is not None:
    cmd += ["--surface_size", str(surface_size[0]), str(surface_size[1])]
if spacing_range is not None:
    cmd += ["--spacing_range", str(spacing_range[0]), str(spacing_range[1])]
cmd += ["--grid", str(grid)]
if grid_range is not None:
    cmd += ["--grid_range", str(grid_range[0]), str(grid_range[1])]

cmd += ["--orientation", str(orientation)]
if orientation_range is not None:
    cmd += ["--orientation_range", str(orientation_range[0]), str(orientation_range[1])]
cmd += ["--lattice_angle", str(lattice_angle)]
if lattice_angle_range is not None:
    cmd += ["--lattice_angle_range", str(lattice_angle_range[0]), str(lattice_angle_range[1])]

cmd += ["--ypc2_r0", str(ypc2_r0)]
cmd += ["--ypc2_sig_r", str(ypc2_sig_r)]
cmd += ["--ypc2_ang_mix", str(ypc2_ang_mix)]
cmd += ["--ypc2_center_plateau_amp", str(ypc2_center_plateau_amp)]
cmd += ["--ypc2_center_plateau_radius", str(ypc2_center_plateau_radius)]
cmd += ["--ypc2_center_plateau_sigma", str(ypc2_center_plateau_sigma)]

if adatom_count is not None:
    cmd += ["--adatom_count", str(adatom_count[0]), str(adatom_count[1])]
if vacancy_count is not None:
    cmd += ["--vacancy_count", str(vacancy_count[0]), str(vacancy_count[1])]
if molecule_count is not None:
    cmd += ["--molecule_count", str(molecule_count[0]), str(molecule_count[1])]
if step_probability is not None:
    cmd += ["--step_probability", str(step_probability)]
if step_height_layers is not None:
    cmd += ["--step_height_layers", str(step_height_layers)]
if step_mode is not None:
    cmd += ["--step_mode", str(step_mode)]
if step_corner_angle_range is not None:
    cmd += ["--step_corner_angle_range", str(step_corner_angle_range[0]), str(step_corner_angle_range[1])]
if step_corner_round_radius is not None:
    cmd += ["--step_corner_round_radius", str(step_corner_round_radius)]
if roughness_sigma is not None:
    cmd += ["--roughness_sigma", str(roughness_sigma)]

if adatom_on_top is not None:
    cmd += ["--adatom_on_top", str(adatom_on_top[0]), str(adatom_on_top[1])]
cmd += ["--adatom_on_top_height", str(adatom_on_top_height)]
if adatom_on_top_offset is not None:
    cmd += ["--adatom_on_top_offset", str(adatom_on_top_offset)]
if adatom_on_top_jitter is not None:
    cmd += ["--adatom_on_top_jitter", str(adatom_on_top_jitter)]
if adatom_on_top_width is not None:
    cmd += ["--adatom_on_top_width", str(adatom_on_top_width)]

if island_count is not None:
    cmd += ["--island_count", str(island_count[0]), str(island_count[1])]
if island_min_distance is not None:
    cmd += ["--island_min_distance", str(island_min_distance)]
if island_orientation_range is not None:
    cmd += ["--island_orientation_range", str(island_orientation_range[0]), str(island_orientation_range[1])]
if island_lattice_angle_range is not None:
    cmd += ["--island_lattice_angle_range", str(island_lattice_angle_range[0]), str(island_lattice_angle_range[1])]
if edge_remove_count is not None:
    cmd += ["--edge_remove_count", str(edge_remove_count[0]), str(edge_remove_count[1])]
if interior_remove_count is not None:
    cmd += ["--interior_remove_count", str(interior_remove_count[0]), str(interior_remove_count[1])]
if edge_remove_edge_prob is not None:
    cmd += ["--edge_remove_edge_prob", str(edge_remove_edge_prob)]

if realistic:
    cmd += ["--realistic"]

if vibration_amp is not None:
    cmd += ["--vibration_amp", str(vibration_amp[0]), str(vibration_amp[1])]
if vibration_wavelength is not None:
    cmd += ["--vibration_wavelength", str(vibration_wavelength[0]), str(vibration_wavelength[1])]
if vibration_angle is not None:
    cmd += ["--vibration_angle", str(vibration_angle[0]), str(vibration_angle[1])]
if slope_x is not None:
    cmd += ["--slope_x", str(slope_x[0]), str(slope_x[1])]
if slope_y is not None:
    cmd += ["--slope_y", str(slope_y[0]), str(slope_y[1])]

if tip_mode is not None:
    cmd += ["--tip_mode", str(tip_mode)]
if tip_sigma is not None:
    cmd += ["--tip_sigma", str(tip_sigma[0]), str(tip_sigma[1])]
if tip_instability is not None:
    cmd += ["--tip_instability", str(tip_instability[0]), str(tip_instability[1])]
if tip_count is not None:
    cmd += ["--tip_count", str(tip_count[0]), str(tip_count[1])]
if tip_offset_range is not None:
    cmd += ["--tip_offset_range", str(tip_offset_range[0]), str(tip_offset_range[1])]
if tip_z_range is not None:
    cmd += ["--tip_z_range", str(tip_z_range[0]), str(tip_z_range[1])]
if tip_weight_range is not None:
    cmd += ["--tip_weight_range", str(tip_weight_range[0]), str(tip_weight_range[1])]
if tip_z_decay is not None:
    cmd += ["--tip_z_decay", str(tip_z_decay)]

print("Running:", " ".join(cmd))
subprocess.run(cmd, check=True, cwd=str(repo_root))
print("Done. Output in", Path(out_dir).resolve())
```
